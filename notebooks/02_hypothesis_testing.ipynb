```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("../data/insurance_data.csv.txt", sep="|")

df["LossRatio"] = df["TotalClaims"] / (df["TotalPremium"] + 1)

# Gender hypothesis test
male = df[df["Gender"] == "Male"]["TotalClaims"]
female = df[df["Gender"] == "Female"]["TotalClaims"]

t_stat, p_value = ttest_ind(male, female, nan_policy="omit")

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

# Hypothesis Testing Analysis

## Objective
The purpose of this analysis is to statistically determine whether insurance risk differs across customer groups and geographic regions for AlphaCare Insurance Solutions (ACIS).

## Business Context
ACIS aims to optimize pricing strategies and identify low-risk customer segments in the South African insurance market. Statistical hypothesis testing helps validate whether observed differences in claims and profitability are significant enough to justify pricing adjustments.

## Key Metrics
The following metrics were used throughout the analysis:

- **Claim Frequency** = proportion of policies with claims
- **Claim Severity** = average claim amount
- **Margin** = TotalPremium − TotalClaims
- **Loss Ratio** = TotalClaims / TotalPremium

## Hypotheses Tested

### H₀₁:
There are no significant risk differences across provinces.

### H₀₂:
There are no significant risk differences between zip codes.

### H₀₃:
There is no significant profit margin difference between zip codes.

### H₀₄:
There is no significant risk difference between men and women.

## Statistical Methods
The following statistical methods were used:

- Independent t-tests
- Descriptive comparison of group means
- P-value evaluation with significance level α = 0.05

## Expected Outcome
The analysis aims to identify customer segments or regions with significantly different insurance risks, supporting data-driven pricing and marketing decisions.
